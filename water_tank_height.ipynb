```python
height=[1,8,7,3,4,2,9]
i=0
j=len(height)-1

max_area=0
while i<j:
  y=j-i
  c=y*min(height[i],height[j])

  if max_area<c:
    max_area=c

  if height[i]<height[j]:
      i=i+1
  else:
      j=j-1
print(max_area,height[i],height[j])
```
