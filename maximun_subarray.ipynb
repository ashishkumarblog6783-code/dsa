```python
def max_sub_array(nums):
    current_sum=nums[0]
    max_sum=nums[0]
    for x in nums[1:]:
      current_sum=max(x,current_sum+x)
      max_sum=max(max_sum,current_sum)
    return max_sum
numbers=[-2,3,1,-8,4,5]
print(max_sub_array(numbers))
```
