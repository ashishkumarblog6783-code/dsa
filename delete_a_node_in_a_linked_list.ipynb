```python
class Node:
  def __init__(self,data):
    self.data=data
    self.next=None

head=Node(10)
second=Node(20)
third=Node(30)

head.next=second
second.next=third

current=head

while current!=None:
  if current.next.data==20:
    current.next=current.next.next
    break

current=current.next
current=head
while current!=None:
    print(current.data)
    current=current.next
```
