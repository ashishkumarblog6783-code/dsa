```python
a="abca"
i=0
j=len(a)-1

while i <j:
  if a[j]==a[j]:
    i+=1
    j-=1
  else:
    b=a[:i]+a[i+1:]
    c=a[:j]+a[j+1:]
    if b==b[::-1] or c ==c[::-1]:
      print(True)
    else:
      print(False)
      break
else:
  print(True)
```
