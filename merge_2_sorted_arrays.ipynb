```python
a=[1,4,7,9]
b=[2,3,5,6]

for i in range(len(a)):
  if a[i]>b[0]:
    a[i],b[0]=b[0],a[i]


    j=0
    if b[j]>b[j+1]:
      b[j],b[j+1]=b[j+1],b[j]
      j+=1
print(sorted(a))
print(sorted(b))
```
