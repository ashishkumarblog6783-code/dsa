```python
def transform_name(name: str) -> str:
    if name.lower().startswith('a'):
        return name.upper()
    return name.lower()

# Examples
names = ["alice", "Bob", "Andrew", "charlie", "adam"]

transformed = [transform_name(n) for n in names]
print(transformed)
# Output: ['ALICE', 'bob', 'ANDREW', 'charlie', 'ADAM']
```
