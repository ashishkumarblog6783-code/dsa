```python
a=[1,1,0,0,0,0,1,1,1]
count=0
for i in range(len(a)-2):
  if a[i]==a[i+1]==a[i+2]:
      count+=1
print(count)
```

```python
a=[1,1,0,0,0,0,1,1,1]
count=0
max_count=1

for i in range(len(a)-1):
  if a[i]==a[i+1]:
    count+=1
  else:
    count=1
  if count>max_count:
    max_count=count

print(max_count)
```
