```python
class  Node:
    def __init__(self,data):
      self.data=data
      self.next=None
head=Node(10)
second=Node(20)
third=Node(30)
fourth=Node(40)
fifth=Node(50)


head.next=second
second.next=third
third.next=fourth
fourth.next=fifth
   2
fifth.next=third

slow=head
fast=head

while fast is not None and fast.next is not None:
  slow=slow.next
  fast=fast.next.next

  if slow==fast:
    print("cycle exist")
    break
else:
      print("not exist")
```
