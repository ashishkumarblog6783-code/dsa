```python
a="abcccd"
target="c"
left=0
right=len(a)-1
while left<=right:
  mid=(left+right)//2
  print("left=",left,"mid=",mid,"right=",right)
  if a[mid]==target:
    print(mid)
    break
  elif a[mid]<target:
    left=mid+1
  else:

    right=mid-1
```
