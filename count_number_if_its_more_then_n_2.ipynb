```python
a=[2,2,2,2,2,1,1,1,1]
l=len(a)/2
for i in range(len(a)):
  count=0
  for j in range(len(a)):
    if a[i]==a[j]:
      count+=1
  if count>l:
    print(a[i])
    break
```
