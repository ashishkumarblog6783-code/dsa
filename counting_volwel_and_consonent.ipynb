```python
a="i am looking"
vowel_count=0
consonent_count=0
v=['a','e','i','o','u']
for i in range(len(a)):
  if a[i]==" ":
    continue
  if a[i] in v :
    vowel_count+=1
  else:
    consonent_count+=1
    print("vowels",vowel_count)

    print("consonent",consonent_count)
```
