```python
a=["a","e","i","o","u","j","p","w"]
count=0
can_count=0
for i in range(len(a)):
  if a[i] in ("a","e","i","o","u") :
    count+=1
    print(a[i],count)

  else:
    can_count+=1
    print("consonents",a[i],can_count)
print(count)
print(can_count)
```
