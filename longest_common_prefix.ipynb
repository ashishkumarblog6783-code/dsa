```python
a="flower","flow","floght"
answer=""
for i in range(len(a[0])):
  for  word in a:
    if i>=len(word):
      break
    if word[i]!=a[0][i]:
      break
  else:
      answer+=a[0][i]
      continue
  break
print(answer)
```
