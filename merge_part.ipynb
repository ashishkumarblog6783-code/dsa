```python
a=[1,3,5,0,0]
b=[2,4]
for i in range(len(b)):
    a.append(b)
while 0 in a:
  a.remove(0)
for i in range(len(a)):
  for j in range(i+1,len(a)):
    if a[i] > a[j]:
      temp=a[i]
      a[i]=a[j]
      a[j]=temp
print(a)
```
