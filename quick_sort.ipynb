```python
a=[3,5,1,8]
pivot=a[-1]
left=[]
right=[]
result=[]
for i in range(len(a)-1):
  if a[i]<pivot:
    left.append(a[i])
  else:
    right.append(a[i])

result=left+[pivot]+right
print(result)
```

```python
def quick_sort(a):
  if len(a)<=1:
    return a
  pivot=a[-1]
  left=[]
  right=[]

  for i in range(len(a)-1):
    if a[i]<pivot:
      left.append(a[i])
    else:
      right.append(a[i])
  result=quick_sort(left)+[pivot]+quick_sort(right)
  return result
a=[4,2,1,9]
print(quick_sort(a))
```
