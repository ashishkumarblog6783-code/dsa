```python
a="aabab"
longest=""

for i in range(len(a)):
  for j in range(len(a)-1,i-1,-1):
    left=i
    right=j
    flag=True

    while left<right:
      if a[left]==a[right]:
        left+=1
        right-=1
      else:
        flag=False
        break
    if flag:
      if len(a[i:j+1])>len(longest):
        longest=a[i:j+1]
  print(longest)
```
