```python
a="{()}]"
if { "{","}","(",")","[","]" } in a:
  print("matching hai babu")
else:
  print("thik se dekh gandu")
```

```python
a="({[]})"
for char in a:
  if char  not in { "{","}","(",")","[","]" }:
    print("missing bracket babu")
    break
  else:
    print("matching babu")
    break
```
