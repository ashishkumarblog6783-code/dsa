```python
a=[5,4,7,1,2]
i=0
j=i+1
for i in range(len(a)):
  for j in range(i+1,len(a)):
    min_index=i

    if a[j]<a[min_index]:
      min_index=j
      a[i],a[min_index]=a[min_index],a[i]
print(a)
```
