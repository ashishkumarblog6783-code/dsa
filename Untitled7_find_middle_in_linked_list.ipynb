```python
class Node:
  def __init__(self,data):
    self.data=data
    self.next=None
a=Node(10)
b=Node(20)
c=Node(30)
d=Node(40)
e=Node(50)

a.next=b
b.next=c
c.next=d
d.next=e

head=a

slow=head
fast=head

while fast and fast.next:
  slow=slow.next
  fast=fast.next.next
print(slow.data)
```
