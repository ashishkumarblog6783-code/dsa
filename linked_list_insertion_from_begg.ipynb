```python
class Node:
  def __init__(self,data):
    self.data=data
    self.next=None
def insert_beginning(head,data):
  new_node=Node(data)
  new_node.next=head
  head=new_node
  return head

def print_list(head):
    current=head
    while current:
      print(current.data,end='->')
      current=current.next
    print('None')

head=Node(10)
head.next=Node(20)
head.next.next=Node(30)
print_list(head)
head=insert_beginning(head,5)
print_list(head)
```
