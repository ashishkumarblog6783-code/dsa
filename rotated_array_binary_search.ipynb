```python
a=[4,5,6,7,0,1,2]
target=0
left=0
right=len(a)-1

while left<=right:
  middle=(left+right)//2
  print("middle")

  if target==a[middle]:
    print(middle)
    break
  if a[left]<= a[middle]:
    if a[left]<=target < a[middle]:
      right=middle-1
    else:
      left=middle+1
  else:
    if a[middle]<target<=a[right]:
      left=middle+1
    else:
      right=middle-1
```
