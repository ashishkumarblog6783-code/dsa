```python
m=("abab")
pattern=("dog cat dog cat")

q=pattern[0:3]
w=pattern[4:7]
e=pattern[8:11]
f=pattern[12:15]

print(q,w,e,f)

if q==e and w==f:
  print("true")
else:
  print("false")
```

```python
m=("abab")
pattern=("dog cat dog fish")
l=pattern.split()
print(l)
a=l[0]
b=l[1]
c=l[2]
d=l[3]
print(a,b,c,d)

if a==c and b==d:
  print("true")
else:
  print("false")
```
