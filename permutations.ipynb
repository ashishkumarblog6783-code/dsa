```python
arr=[1,2,3]

i=len(arr)-2

while i>=0 and arr[i] >= arr[i+1]:
  i-=1

if i >=0:
  j=len(arr) - 1

  while arr[j] <=arr[i]:
    j-=1

  arr[i],arr[j]=arr[j],arr[i]

  arr[i+1 :]=reversed(arr[i+1 :])

  print(arr)
```

```python
from itertools import permutations
arr=[2,3,4]
result=permutations(arr)
for x in result:
  print(x)
```
