```python
digit="23"

phone={
    "2":"abc",
    "3":"def",
    "4":"ghi",
    "5":"jkl",
    "6":"mno",
    "7":"pqr",
    "8":"stw",
    "9":"xyz"
}
for i in phone[digit[0]]:
  for j in phone[digit[1]]:
    print(i+j)
```
