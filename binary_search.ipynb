```python
a=[2,5,3,9]
target=9
left=0
right=len(a)-1
while left<=right:
  mid=(left+right)//2
  if a[mid]==target:
    print(mid)
    break
  elif a[mid]<target:
    left=mid+1
  else:
    right=mid-1
```
