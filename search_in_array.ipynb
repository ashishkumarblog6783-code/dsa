```python
a=[1,24,0,8,6]
a=sorted(a)
target=24
left=0
right=len(a)-1
while left <=right:

  mid=(left+right)//2

  if target==a[mid]:
    print(mid)
    break
  elif target>a[mid]:
    left=mid+1
  else:
    right=mid-1
```
